```python
import nltk
from nltk.translate.bleu_score import sentence_bleu, SmoothingFunction
from rouge_score import rouge_scorer

# Ensure NLTK resources are downloaded
nltk.download('wordnet')
```

# Evaluating Small Language Models: BLEU, ROUGE, and METEOR

When training or fine-tuning a language model, we need proper evaluation metrics to compare the model’s performance. In this notebook, we explore three common metrics:

- **BLEU (Bilingual Evaluation Understudy)**
- **ROUGE (Recall-Oriented Understudy for Gisting Evaluation)**
- **METEOR (Metric for Evaluation of Translation with Explicit ORdering)**

Each metric has its own strengths and is used in different scenarios.

```python
### Example Sentences for Evaluation
reference = "The cat sat on the mat."
hypothesis_1 = "The cat is sitting on the mat."
hypothesis_2 = "A dog sat on the floor."
```

## 1. BLEU (Bilingual Evaluation Understudy)

BLEU measures how many words (n-grams) in the generated text overlap with the reference text.
- It is mainly used in **machine translation** and **text generation**.
- It works well when the output needs to be very **precise** and **grammatically similar**.
- BLEU ignores meaning, so synonyms may not score well.


```python
# BLEU Calculation
smoothie = SmoothingFunction().method1
bleu_score = sentence_bleu([reference.split()], hypothesis_1.split(), smoothing_function=smoothie)
print("BLEU Score:", bleu_score)
```

## 2. ROUGE (Recall-Oriented Understudy for Gisting Evaluation)

ROUGE is often used for **summarization tasks** as it focuses on recall rather than precision.
- **ROUGE-1**: Overlap of unigrams (single words)
- **ROUGE-2**: Overlap of bigrams (two consecutive words)
- **ROUGE-L**: Longest Common Subsequence (LCS), useful for long-form text.

ROUGE is useful when the **essence** of the generated text matters more than exact words.

```python
# ROUGE Calculation
scorer = rouge_scorer.RougeScorer(['rouge1', 'rouge2', 'rougeL'], use_stemmer=True)
rouge_scores = scorer.score(reference, hypothesis_1)
print("ROUGE Scores:", rouge_scores)
```

## 3. METEOR (Metric for Evaluation of Translation with Explicit ORdering)

METEOR is another metric used in **machine translation and summarization**. It is better than BLEU in handling:
- Synonyms
- Word ordering
- Morphological variations

It is a more **human-aligned** metric but computationally more expensive.

```python
from nltk.translate.meteor_score import meteor_score

# METEOR Calculation
meteor = meteor_score([reference.split()], hypothesis_1.split())
print("METEOR Score:", meteor)
```

## Summary
- **BLEU**: Best for precise text generation like **machine translation**.
- **ROUGE**: Best for **text summarization** where capturing **important words** matters.
- **METEOR**: More **human-aligned**, better at **synonym handling**, useful for **summarization and translation**.

By using a combination of these metrics, we can get a **holistic** evaluation of a language model’s performance.
