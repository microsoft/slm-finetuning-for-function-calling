### ONNX Runtime 

ONNX is an open format built to represent machine learning models.  It provides interoperability by supporting various deep-learning frameworks, allowing models to be converted from frameworks such as PyTorch and TensorFlow, to the ONNX format.  Models can then be run using the ONNX Runtime which is compatible with different hardware, drivers, and operating systems, and provides optimal performance by leveraging hardware accelerators where applicable alongside graph optimizations and transforms

✅ Why Use It?
- Cross-platform.
- Optimized for CPUs and low-power devices.

![ONNX](images/onnx.png)

For engineers or data-scientists, the value proposition is "train your models in your favourite framework without worrying about inference-time constraints. After training, just convert your model to ONNX format".

However, you still need to create the model first, before converting to the ONNX format first. Or you can download a model from the ONNX model zoo.  https://github.com/onnx/models 


## 🤗 Hugging Face 

Hugging Face is a company and open-source community which makes AI more accessible. 

It provides:
1. **Transformers**: an open-source Python library that provides access to thousands of pre-trained Transformer models for natural language processing (NLP), computer vision, audio tasks.
2. **Model Hub**: a repository with over 900k models, 200k datasets, and 300k demo apps (Spaces), all open source and publicly available, in an online platform where people can easily collaborate and build ML together
3. **Datasets**: a library for accessing and sharing datasets (not just text)for audio, computer vision, and natural language processing (NLP) tasks.  The datasets library has utilities for reading datasets from the Hugging Face Hub. 
4. **Inference API**: allowing Pro users to run models without downloading them locally.
   https://huggingface.co/docs/api-inference/en/index 

Hugging Face is popular because it makes AI accessible with pre-trained models, standardizes NLP workflows, and provides an active community.

## How do you choose which small language model to use? 

🤔 With many language models to choose from, what decision drivers or criteria would you use to choose between them?

Some criteria: 

- The task - Is the model capable of doing what you want it to do?

- The size of the model - Do you have enough compute to handle it?
    - Tiny (under 1B parameters) → Fastest, good for mobile/edge devices.
	- Small (1B - 7B) → Good balance between performance and efficiency.
	- Medium (7B - 13B) → Higher accuracy, needs GPUs.
    Tip: If you're deploying on CPU or mobile

- Performance trade off - some SLMs have higher latency, or perform better in benchmarks

- Context length 

- The cost - Is it too expensive to run?

- The license - Is it suitable for commercial use?
	
- Maintainer credibility and long-term support
